# 🛡️ Atividade 4: O Grande Registro da Guilda

A nossa guilda está crescendo! O Mestre da Guilda (Professor) criou um repositório central para guardar os registros de todos os aventureiros.

Sua missão é clonar este repositório, criar uma pasta com o seu nome e preencher sua ficha de personagem e seus equipamentos. No final, vamos sincronizar tudo para que todos tenham acesso ao catálogo completo da equipe!

### 📥 Passo 1: Clonar o Repositório do Professor

1. O professor enviará o link do repositório no GitHub.
2. No **VS Code**, pressione `F1` (ou `Ctrl+Shift+P`), digite `Git: Clone` e dê Enter.
3. Cole a URL do professor e escolha uma pasta no seu computador para salvar.
4. Clique em **Open** (Abrir) quando o VS Code perguntar.

### 📁 Passo 2: Criando sua Ficha e Equipamentos

1. No menu lateral do VS Code (Explorer), clique no ícone de **New Folder** (Nova Pasta).
2. Dê à pasta o seu **Nome e Sobrenome** (ex: `raphael_rocha`).
3. Clique com o botão direito **em cima da sua pasta** e crie um arquivo chamado `personagem.md`.
4. Dentro de `personagem.md`, coloque sua classe e atributos:
   - **Nome:** 
   - **Classe:** (Ex: Mago do Python, Guerreiro do Front-end)
   - **Habilidade Principal:** 
5. Salve o arquivo (`Ctrl+S`).
6. Crie outro arquivo **dentro da sua pasta** chamado `equipamentos.md`.
7. Liste 3 itens que você carrega na mochila (ex: Teclado Mecânico +3, Poção de Café, etc.).
8. Salve o arquivo (`Ctrl+S`).

### 💾 Passo 3: Preparar e Salvar (Add e Commit)

Vamos registrar sua pasta nova no histórico do Git.

1. Abra o terminal integrado do VS Code (`Terminal > New Terminal`).
2. Adicione sua pasta e arquivos com o comando:
   `git add .`
3. Salve com uma mensagem clara:
   `git commit -m "Adicionando ficha e equipamentos de [Seu Nome]"`

### 🚀 Passo 4: O Engarrafamento na Nuvem (Push e Pull)

Agora você precisa enviar sua pasta para o repositório do professor.

1. No terminal, digite:
   `git push origin main`

**⚠️ ATENÇÃO - DEU ERRO (REJECTED)?**
Isso significa que outro aluno foi mais rápido e enviou a ficha dele antes de você. O Git não deixa você enviar até que baixe o que o colega fez!

**Como resolver:**
1. Baixe a pasta do seu colega com o comando: 
   `git pull origin main`
*(Se abrir uma tela estranha de texto no terminal, digite `:wq` e aperte Enter para confirmar a mesclagem automática das pastas).* 
2. Agora que seu computador tem a sua pasta e a do seu colega, tente enviar de novo:
   `git push origin main`

### 🔄 Passo 5: O Ponto de Encontro Final

Quando o professor avisar que **todos os alunos** conseguiram fazer o Push, o repositório na nuvem estará cheio de pastas.

Mas o seu VS Code ainda não está vendo os colegas que enviaram os arquivos depois de você.

1. Para ter o catálogo completo da guilda no seu computador, rode o comando uma última vez:
   `git pull origin main`

**🎉 Parabéns!** Olhe para a esquerda no Explorer: você verá as pastas e as fichas de todos os seus colegas de classe!
